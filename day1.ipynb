```python
with open("input.txt", "r") as file:
    content = file.read()
    inp = content
```

```python
def clean_instructions(s):
    arr = s.split("\n")
    res = [(i[0], int(i[1:])) for i in arr if len(i) > 1]
    return res
```

## Part 1

```python
def num_zero(instructions):
    count_zero = 0
    sum = 50
    for dir, val in instructions:
        if dir == 'L':
            sum -= val
        else:
            sum += val
        sum %= 100
        if sum == 0:
            count_zero += 1
        # print(dir, val, sum)
    return count_zero
```

```python
num_zero(clean_instructions(example_input))
```

```python
num_zero(clean_instructions(inp))
```

### Solving with Example Case

```python
def num_clicked_zero(instructions):
    count_zero = 0
    sum =prev= 50
    for dir, val in instructions:
        prev = sum
        sum += val//100
        if dir == "L":
            sum -= val
        else:
            sum += val
            
        if prev != 0 and sum <= 0 or sum >= 100 :
            count_zero += 1

        # print(dir, val, sum, count_zero)
        sum %= 100
    return count_zero +1 if sum == 0 else count_zero
```

```python
num_clicked_zero(clean_instructions(example_input))
```

```python
num_clicked_zero(clean_instructions(inp))
```
