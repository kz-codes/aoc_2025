```python
example_input ="""
L68
L30
R48
L5
R60
L55
L1
L99
R14
L82
"""
example_input
```

```python
with open("input.txt", "r") as file:
    content = file.read()
    inp = content
```

```python
def clean_instructions(s):
    arr = s.split("\n")
    res = [(i[0], int(i[1:])) for i in arr if len(i) > 1]
    return res
```

## Part 1

```python
def num_zero(instructions):
    count_zero = 0
    pos = 50
    for dir, val in instructions:
        if dir == 'L':
            pos -= val
        else:
            pos += val
        pos %= 100
        if pos == 0:
            count_zero += 1
        # print(dir, val, pos)
    return count_zero
```

```python
num_zero(clean_instructions(example_input))
```

```python
num_zero(clean_instructions(inp))
```

## Part 2

### Solving with Example Case

```python
def num_clicked_zero(instructions):
    count_zero = 0
    pos = 50
    for dir, val in instructions:
        for _ in range(val):
            if dir == 'R':
                pos += 1
            else:
                pos -= 1
            
            if pos == -1 or pos == 100:
                pos %= 100
            
            if pos == 0:
                count_zero += 1
    return count_zero
```

```python
num_clicked_zero(clean_instructions(example_input))
```

```python
num_clicked_zero(clean_instructions(inp))
```
