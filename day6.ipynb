```python
example_input = """
123 328  51 64  45 64  387 23 6 98  215 314
*   +   *   + 
"""
```

```python
with open('input.txt', 'r') as filename:
    inp = filename.read()
```

```python
def parse(input):
    res = []
    ops = []
    l,r = 0,0
    while r < len(input):
        if input[l].isnumeric():
            while input[r].isnumeric():
                r += 1
            res.append(input[l:r])
            l = r
            continue
        elif input[l] in '*+':
            ops.append(input[l])
            
        l += 1
        r += 1
    
    
    
    return res,ops
```

```python
from sspipe import p
example_input | p(parse)
```

## Part 1

```python
def part1(data):
    arr = data[0]
    ops = data[1]
    step = len(ops)
    total = 0
    for offset,op in enumerate(ops):
        p = 0
        if op == '*':
            p = 1
            for i in range(offset, len(arr), step):
                p *= int(arr[i])
        
        if op == '+':
            for i in range(offset, len(arr), step):
                p += int(arr[i])
        total += p
    return total
            
```

```python
inp | p(parse) |p(part1)
```
