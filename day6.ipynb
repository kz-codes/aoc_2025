```python
example_input = """
123 328  51 64  45 64  387 23 6 98  215 314
*   +   *   + 
"""
```

```python
with open("input.txt", "r") as filename:
    inp = filename.read()
```

```python
def parse(input):
    res = []
    ops = []
    l, r = 0, 0
    while r < len(input):
        if input[l].isnumeric():
            while input[r].isnumeric():
                r += 1
            res.append(input[l:r])
            l = r
            continue
        elif input[l] in "*+":
            ops.append(input[l])

        l += 1
        r += 1

    return res, ops
```

```python
from sspipe import p

example_input | p(parse)
```

## Part 1


```python
def part1(data):
    arr = data[0]
    ops = data[1]
    step = len(ops)
    total = 0
    for offset, op in enumerate(ops):
        p = 0
        if op == "*":
            p = 1
            for i in range(offset, len(arr), step):
                p *= int(arr[i])

        if op == "+":
            for i in range(offset, len(arr), step):
                p += int(arr[i])
        total += p
    return total
```

```python
inp | p(parse) | p(part1)
```

## Part 2


```python
def part2(text):
    # Split into lines and pad them to form a perfect grid
    lines = text.splitlines()
    if not lines:
        return 0
    width = max(len(line) for line in lines)
    grid = [line.ljust(width) for line in lines]
    H = len(grid)

    # 1. Identify "Problem Blocks"
    # Problems are separated by columns that are entirely spaces.
    # We find the column indices that belong to problems.
    blocks = []
    current_block = []

    for x in range(width):
        # Check if column x is a separator (all spaces)
        is_separator = all(grid[y][x] == " " for y in range(H))

        if not is_separator:
            current_block.append(x)
        elif current_block:
            blocks.append(current_block)
            current_block = []

    if current_block:
        blocks.append(current_block)

    total = 0

    # 2. Process each block
    for cols in blocks:
        # The operator is located in the last row of the block [web:1]
        op = None
        for x in cols:
            char = grid[H - 1][x]
            if char in "+*":
                op = char
                break

        if not op:
            continue

        # Extract numbers vertically from each column in the block
        # We read from top to bottom, excluding the last row (operator row)
        nums = []
        for x in cols:
            # "Compact" the column: join chars and remove spaces
            col_chars = [grid[y][x] for y in range(H - 1)]
            num_str = "".join(col_chars).replace(" ", "")

            if num_str:
                nums.append(int(num_str))

        # 3. Calculate result for this block
        if op == "+":
            total += sum(nums)
        elif op == "*":
            res = 1
            for n in nums:
                res *= n
            total += res

    return total

```

```python
inp | p(part2)
```
