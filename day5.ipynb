```python
from sspipe import p
def parse(input):
    res = {"ranges": [], "ids": []}
    ll = input.split('\n\n')
    ranges = [e for e in ll[0].split('\n') if len(e) > 1]
    # ids = [e for e in ll[0] if len(e) > 1]
    for r in ranges:
        lp = r.split('-')
        e = int(lp[0]), int(lp[1])
        res['ranges'].append(e)
    ids = [int(e) for e in ll[1].split('\n') if len(e) >= 1]
    for id in ids:
        res["ids"].append(id)
    return res
```

```python
def num_fresh(data):
    ids = data['ids']
    ranges = data['ranges']
    
    count = 0
    for id in ids:
        for a,b in ranges:
            if a <= id <= b:
                count += 1
                break
    return count
```
