```python
example_input = """
3-5
10-14
16-20
12-18

1
5
8
11
17
32
"""
```

```python
with open("input.txt", "r") as file:
    inp = file.read()
```

```python
from sspipe import p


def parse(input):
    res = {"ranges": [], "ids": []}
    ll = input.split("\n\n")
    ranges = [e for e in ll[0].split("\n") if len(e) > 1]
    # ids = [e for e in ll[0] if len(e) > 1]
    for r in ranges:
        lp = r.split("-")
        e = int(lp[0]), int(lp[1])
        res["ranges"].append(e)
    ids = [int(e) for e in ll[1].split("\n") if len(e) >= 1]
    for id in ids:
        res["ids"].append(id)
    return res
```

## Part 1


```python
def num_fresh(data):
    ids = data["ids"]
    ranges = data["ranges"]

    count = 0
    for id in ids:
        for a, b in ranges:
            if a <= id <= b:
                count += 1
                break
    return count
```

```python
example_input | p(parse) | p(num_fresh)
```

```python
inp | p(parse) | p(num_fresh)
```

## Part 2


```python
def optimized_ranges(ranges):
    lows, highs = [], []
    for a, b in ranges:
        lows.append(a)
        highs.append(b)
    lows.sort()
    highs.sort()

    res = []
    l, r = 0, 0

    while l < len(lows):
        if r + 1 < len(lows) and highs[r] >= lows[r + 1]:
            r += 1
        else:
            res.append((lows[l], highs[r]))
            r += 1
            l = r
    return res
```

```python
def num_fresh_ranges(data):
    ranges = data["ranges"]
    ranges = optimized_ranges(ranges)

    total = 0
    for a, b in ranges:
        total += b - a + 1
    return total
```

```python
example_input | p(parse) | p(num_fresh_ranges)
```

```python
inp | p(parse) | p(num_fresh_ranges)
```
