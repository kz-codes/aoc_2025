```python
def parse(text):
    return text.split('\n')
```

```python
from sspipe import p
example_input | p(parse)
```

## Part 1

```python
def num_splits(data):
    rays = [False] * len(data[0])
    splits = 0
    for row in data:
        for i, e in enumerate(row):
            if e == 'S':
                rays[i] = True
            if e == '^' and rays[i]:
                rays[i] = False
                rays[i-1] = True
                rays[i+1] = True
                splits+=1
    return splits
```

```python
example_input | p(parse) | p(num_splits)
```

```python
inp | p(parse) | p(num_splits)
```

## Part 2

```python
def num_possible_timeline(data):
    width = len(data[0])
    rays = [0]*width
    
    rays[data[0].index('S')] = 1
    
    for r in range(1,len(data)):
        row = data[r]
        new_rays = [0] * width
        
        for i,char in enumerate(row):
            count = rays[i]
            if count == 0:
                continue
            
            if char == '^':
                new_rays[i-1] += count
                new_rays[i+1] += count
            elif char == '.':
                new_rays[i] += count
        rays = new_rays
    return sum(rays)
```

```python
example_input | p(parse) | p(num_possible_timeline)
```

```python
inp | p(parse) | p(num_possible_timeline)
# inp | p(parse) | p(len) 
```
