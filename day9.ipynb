## Part 1


```python
def maxArea(data):
    max_area = 0
    for i in range(len(data)):
        for j in range(i + 1, len(data)):
            p = data[i]
            q = data[j]
            area = abs((p[0] - q[0] + 1) * (p[1] - q[1] + 1))
            # print(area,max_area, p,q)
            max_area = max(max_area, area)
    return max_area
```

```python
example_input | p(parse) | p(maxArea)
```

```python
from collections import defaultdict


def maxArea_2(data):
    # Existing points
    point_set = set(data)

    # points sorted by x
    points = sorted(data)
    n = len(points)

    # for a x where in y has red tile
    cols = defaultdict(list)
    for x, y in points:
        cols[x].append(y)

    # all x where red tile exists
    xs = sorted(cols.keys())

    max_area = 0
    for i in range(n):
        px, py = points[i]
        for j in range(i + 1, n):
            qx, qy = points[j]

            if px == qx or py == qy:
                continue

            # if (px, qy) not in point_set or (qx, py) not in point_set:
            #     continue

            area = (qx - px + 1) * (abs(py - qy) + 1)
            if area <= max_area:
                continue

            is_valid = True
            ly, uy = min(py, qy), max(py, qy)

            # checks if any red tile in x coord
            for x in xs:
                if x <= px:
                    continue
                if x >= qx:
                    break

                # checks in y coord
                for y in cols[x]:
                    if ly < y < uy:
                        is_valid = False
                        break
                if not is_valid:
                    break
            if is_valid:
                max_area = area
                print(points[i], points[j], max_area)
    return max_area
```

```python
example_input | p(parse) | p(maxArea_2)
```
