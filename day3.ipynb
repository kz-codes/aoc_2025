```python
example_input = """
987654321111111
811111111111119
234234234234278
818181911112111
"""
```

```python
with open("input.txt", "r") as file:
    content = file.read()
    inp = content
```

```python
def clean_input(inp):
    ll = [s for s in inp.split("\n") if len(s) > 1]
    return [[int(e) for e in row] for row in ll]
```

```python
from sspipe import p, px

example_input | p(clean_input)
```

## Part 1


```python
def max_joltage(bank):
    left = 0
    for i in range(len(bank) - 1):
        if bank[i] > bank[left]:
            left = i
    right = left + 1
    for i in range(left + 1, len(bank)):
        if bank[i] > bank[right]:
            right = i

    return bank[left] * 10 + bank[right]


def part1(ll):
    sum = 0
    for row in ll:
        sum += max_joltage(row)
    return sum
```

```python
example_input | p(clean_input) | p(part1)
```

```python
inp | p(clean_input) | p(part1)
```

## Part 2


```python
def get_joltage(bank: list[int], num_digits: int):
    batteries = [-1]
    for b in range(1, num_digits + 1):
        batteries.append(batteries[b - 1] + 1)

        for i in range(batteries[b], len(bank) - num_digits + b):
            if bank[i] > bank[batteries[b]]:
                batteries[b] = i

    on_batteries = [str(bank[i]) for i in batteries[1:]]
    return int("".join(on_batteries))


def day2(ll):
    total = 0
    for bank in ll:
        total += get_joltage(bank, 12)
    return total
```

```python
inp | p(clean_input) | p(day2)
```
