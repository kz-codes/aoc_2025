## Part 1


```python
import heapq
import itertools
import math


def smallest_edges(points, num_of_edges):
    edges = [
        (math.dist(p1, p2), p1, p2) for p1, p2 in itertools.combinations(points, 2)
    ]
    return heapq.nsmallest(num_of_edges, edges)
```

```python
class UnionFind:
    def __init__(self) -> None:
        self.parent = {}

    def find(self, i):
        if i not in self.parent:
            self.parent[i] = i

        if self.parent[i] != i:
            self.parent[i] = self.find(self.parent[i])
        return self.parent[i]

    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i != root_j:
            self.parent[root_i] = root_j
            return True
        return False
```

```python
from collections import Counter


def part1(nodes):
    uf = UnionFind()

    for _, p1, p2 in nodes:
        uf.union(p1, p2)

    group_counts = Counter()
    for p in uf.parent:
        root = uf.find(p)
        group_counts[root] += 1

    t3 = group_counts.most_common(3)

    p = 1
    for _, size in t3:
        p *= size

    return p
```

```python
inp | p(parse) | p(smallest_edges, 1000) | p(part1)
```

## Part 2


```python
class UnionFind2:
    def __init__(self, points) -> None:
        self.parent = {p: p for p in points}
        self.count = len(points)

    def find(self, i):
        if i not in self.parent:
            self.parent[i] = i

        if self.parent[i] != i:
            self.parent[i] = self.find(self.parent[i])
        return self.parent[i]

    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)

        if root_i != root_j:
            self.parent[root_i] = root_j
            self.count -= 1
            return True
        return False
```

```python
def sorted_edges(points):
    return sorted(
        [(math.dist(p1, p2), p1, p2) for p1, p2 in itertools.combinations(points, 2)]
    )
```

```python
def part2(points):
    uf = UnionFind2(points)
    edges = sorted_edges(points)
    lp1, lp2 = None, None

    for _, p1, p2 in edges:
        if uf.union(p1, p2):
            lp1, lp2 = p1, p2
            if uf.count == 1:
                break
    if lp1 and lp2:
        return lp1[0] * lp2[0]
    return 0
```

```python
inp | p(parse) | p(part2)
```
