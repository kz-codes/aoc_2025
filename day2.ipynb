```python
example_input = "11-22,95-115,998-1012,1188511880-1188511890,222220-222224,1698522-1698528,446443-446449,38593856-38593862,565653-565659,824824821-824824827,2121212118-2121212124"
```

```python
with open("input.txt", "r") as file:
    content = file.read()
    inp = content
```

```python
def clean_input(s):
    return [
        (
            int(i.split("-")[0]),
            int(i.split("-")[1]),
        )
        for i in s.split(',')
        if len(i) > 1
    ]
```

## Part 1

```python
from functools import reduce

def list_invalid_ids(ranges):
    invalid_ids = []
    for a,b in ranges:
        for i in range(a,b+1):
            s = str(i)
            l = len(s)
            if l % 2 == 0:
                if s[:l//2] == s[l//2:]:
                    invalid_ids.append(i)
    return invalid_ids

def solve(ll):
    return reduce(lambda x, y: x + y, ll)
```

```python
from sspipe import p
# solve(list_invalid_ids(clean_input(example_input)))
example_input | p(clean_input) | p(list_invalid_ids) | p(solve)
```

```python
inp | p(clean_input) | p(list_invalid_ids) | p(solve)
```

## Part 2

```python
def is_invalid(s):
    for c in range(1, len(s)//2 + 1):
        if not len(s) % c == 0:
            continue
        
        substrs = [s[i: i+c] for i in range(0, len(s), c)]
        if all(substrs[0] == subs for subs in substrs):
            return True
    return False

def list_invalid_ids_2(ranges):
    res = []
    for a,b in ranges:
        for i in range(a,b+1):
            if is_invalid(str(i)):
                res.append(i)
    return res          
```

```python
example_input | p(clean_input) | p(list_invalid_ids_2) | p(solve)
```

```python
inp | p(clean_input) | p(list_invalid_ids_2) | p(solve)
```
